````python
# Primero normalizamos las columnas.
# Ejecutamos esto ANTES de nada para unificar criterios y para evitar que hayan diferencias entre archivos.
# 1. str.strip(): Elimina espacios en blanco al principio/final.
# 2. str.replace('-', '_'): Cambia guiones medios por bajos (snake_case).
multas.columns = multas.columns.str.strip().str.replace('-', '_')
print("Columnas normalizadas a snake_case:", multas.columns.tolist())

# a)
print(f"\n--- APARTADO A ---")
# Como ya hemos limpiado las columnas arriba, accedemos sin problemas
print(f"Hay {multas['CALIFICACION'].unique().size} valores distintos en la categoria CALIFICACION que son: {multas['CALIFICACION'].unique()}")
multas['CALIFICACION'] = multas['CALIFICACION'].str.strip()


# b)
print(f"\n--- APARTADO B ---")
# Aquí ya usamos 'HECHO_BOL', con guion bajo porque lo hemos normalizado
cols_texto = ['DESCUENTO', 'HECHO_BOL', 'DENUNCIANTE']

for col in cols_texto:
    # Verificamos si la columna existe antes de acceder para evitar errores
    if col in multas.columns:
        multas[col] = multas[col].str.strip()
        print(f"Hay {multas[col].unique().size} valores distintos en {col}")
    else:
        print(f"La columna {col} no se encuentra en el dataframe.")


# c)
print(f"\n--- APARTADO C ---")
print("Columnas actuales:", multas.columns.tolist())

# Bateria de preguntas:

col_vel = [c for c in multas.columns if 'VEL_CIRCULA' in c][0]
col_y = [c for c in multas.columns if 'COORDENADA' in c and 'Y' in c][0]

print(f"Columna que representa la velocidad registrada: '{col_vel}'")
print(f"Columna coordenada Y: '{col_y}'")
print(f"Muestra de la coordenada Y:\n{multas[col_y].head(3)}")

# En el enunciado pediais que usaramos el 'rename' aquí, pero ya he normalizado los nombres de las columans al principio por lo que
# no le he encontrado mucho sentido volverlo a realizar.
# En caso de realizarlo usaria una expresion lambda dentro de la funcion rename aplicandole el strip 'multas.rename(columns=lambda x: x.strip())'


# d)
print(f"\n--- APARTADO D ---")
print(f"Tipo antes de conversión VEL_LIMITE: {multas['VEL_LIMITE'].dtype}")

multas['VEL_LIMITE'] = pd.to_numeric(multas['VEL_LIMITE'], errors='coerce')
multas['VEL_CIRCULA'] = pd.to_numeric(multas['VEL_CIRCULA'], errors='coerce')

print(f"Valores distintos VEL_LIMITE: {multas['VEL_LIMITE'].nunique()}")
print(f"Valores distintos VEL_CIRCULA: {multas['VEL_CIRCULA'].nunique()}")


# e)
print(f"\n--- APARTADO E ---")
# Al igual que antes, usamos el '_' en lugar del '-' en los nombres de las columnas.
print(f"Tipo antes de conversión COORDENADAS: {multas['COORDENADA_X'].dtype}")

multas['COORDENADA_X'] = pd.to_numeric(multas['COORDENADA_X'], errors='coerce')
multas['COORDENADA_Y'] = pd.to_numeric(multas['COORDENADA_Y'], errors='coerce')

print("Tipos finales tras conversiones:")
print(multas.dtypes)
````

### Creación de columna nueva

__a)__ A partir del dataframe `multas` que contiene las columnas `MES`, `ANIO` y `HORA`, crear una nueva columna llamada `fecha` de tipo datetime que combine dicha información en un único objeto de fecha y hora. Usa como día el día 1.

__b)__ Modifica el dataframe `multas` de forma que el índice sea el valor de la columna creada anteriormente.

__c)__ Elimina del dataframe la columna `fecha`.



````python
# Primero vamos a separar horas y minutos de la columna 'HORA' donde la  parte entera es la hora y por si acaso rellenamos nulos con 0 para evitar errores
hora_float = multas['HORA'].fillna(0)
horas = hora_float.astype(int)

# La parte decimal son los minutos (multiplicamos por 100 y redondeamos para evitar errores de coma flotante
# esto fue un consejo de nuestra amiga la IA bastante útil)
# El ejemplo que me puso: 20.30 -> 0.30 * 100 = 30
minutos = ((hora_float - horas) * 100).round().astype(int)

# Ahora creamos la fecha completa usando solo el dia 1
fechas_componentes = {
    'year': multas['ANIO'],
    'month': multas['MES'],
    'day': 1,
    'hour': horas,
    'minute': minutos
}
multas['fecha'] = pd.to_datetime(fechas_componentes)

# Y usamos la función set_index que hace automáticamente las dos cosas por defecto, la establecee como índice y la elimina de la lista de columnas
multas.set_index('fecha', inplace=True)

print("Índice transformado a DatetimeIndex:")
display(multas.head(10))
````

### C7


A partir del dataframe `resumen_500` calculado anteriormente, vamos a representar en un mapa de madrid los puntos donde se ocurren las infracciones. Estudiar y ejecutar el siguiente código:

````python
# C7

from pyproj import Transformer
import folium

# Cogemos una muestra de las multas originales que tengan coordenadas y filtramos las que tienen coordenadas
con_coords = multas[(multas['COORDENADA_X'] > 0) & (multas['COORDENADA_Y'] > 0)].copy()

# Cogemos las 500 más relevantes para no saturar el mapa porque si no saturaba demasiado.
mapa_data = con_coords.sort_values('IMP_BOL', ascending=False).head(500)

# Convertimos las coordenadas UTM (EPSG:25830) a Lat/Lon (EPSG:4326)
transformer = Transformer.from_crs("epsg:25830", "epsg:4326", always_xy=False)

def convertir_coords(row):
    lat, lon = transformer.transform(row['COORDENADA_X'], row['COORDENADA_Y'])
    return pd.Series({'lat': lat, 'lon': lon})

mapa_data[['lat', 'lon']] = mapa_data.apply(convertir_coords, axis=1)

# Creamos el mapa centrado en Madrid
m = folium.Map(location=[40.4167, -3.70325], zoom_start=12)

# Añadimos marcadores con colores según calificación
colores = {'LEVE': 'green', 'GRAVE': 'red', 'MUY GRAVE': 'darkred'}

for idx, row in mapa_data.iterrows():
    tipo = row['CALIFICACION']
    color_punto = colores.get(tipo, 'blue')

    folium.CircleMarker(
        location=[row['lat'], row['lon']],
        radius=5,
        color=color_punto,
        fill=True,
        fill_color=color_punto,
        fill_opacity=0.7,
        popup=f"{row['LUGAR']}<br>{tipo}<br>{row['IMP_BOL']}€"
    ).add_to(m)

m.save('mapa_multas_madrid.html')
print("Mapa generado correctamente: 'mapa_multas_madrid.html'")
m
````

`pyproj`  se usa para trabajar con sistemas de coordenadas geográficas, mientras que `Transformer` permite convertir coordenadas entre diferentes sistemas de referencia (por ejemplo, de coordenadas UTM a latitud/longitud WGS84).

### C12
¿Cuánto dinero se ha recaudado (máximo y mínimo) y cuántas multas se han tramitado   por tipo de tramitador?. Una multa puede tener descuento por pronto pago del 50%, pero cabe la posibilidad de que el infractor no realice el pago en fecha para acojerse a dicho descuento.

````python
# C12
# Explicación de la solución en base a la tutoria con Yolanda
# - Recaudación Máxima: Suma de IMP_BOL (asumiendo que nadie paga pronto).
# - Recaudación Mínima: Si tienen derecho a descuento ('SI'), asumimos que pagan el 50%.
#                       Si NO tienen derecho ('NO'), pagan el 100%.

# Calculamos una columna temporal con el importe mínimo teórico de cada multa
multas['imp_minimo_teorico'] = np.where(
    multas['DESCUENTO'] == 'SI',
    multas['IMP_BOL'] * 0.5,
    multas['IMP_BOL']
)

# Agrupamos por DENUNCIANTE
resumen_recaudacion = multas.groupby('DENUNCIANTE').agg({
    'IMP_BOL': ['count', 'sum'],         # Count = num_multas, Sum = Recaudación Máxima
    'imp_minimo_teorico': 'sum'          # Sum = Recaudación Mínima calculada
})

# Aplanamos y renombramos las columnas para que quede limpio
resumen_recaudacion.columns = ['num_multas', 'recaudacion_max', 'recaudacion_min']

# Ordenamos por recaudación máxima
resumen_recaudacion = resumen_recaudacion.sort_values('recaudacion_max', ascending=False)

# Visualización
print("Resumen de recaudación por tramitador:")
display(resumen_recaudacion.style.format({
    'recaudacion_max': '{:,.2f} €',
    'recaudacion_min': '{:,.2f} €'
}))

# Borramos la columna temporal para dejar el DF como estaba
multas.drop(columns=['imp_minimo_teorico'], inplace=True)
````

### Clase `Cache`

Se encargará de guardar y recuperar datos en disco, de forma que estén disponibles incluso después de cerrar el programa.

La clase `Cache` va a almacenar cualquier tipo de fichero asociado con un identificador. Por defecto, los datos se almacenarán en el home del usuario, en un directorio para este ejercicio. Por ejemplo:
```
CACHE_DIR = Path.home() / ".my_cache"
```
Consultar [Path.home](https://docs.python.org/3/library/pathlib.html#pathlib.Path.home) del módulo `pathlib`.
Pero en el constructor se podría indicar que fuera cualquier otra carpeta del sistema operativo.

La clase `Cache` debe estar pensada para ser usada desde otras aplicaciones, no solo este ejercicio, por lo que cada objeto de la clase cache deberá tener los siguientes atributos:

* `app_name`, un `str` que será un nombre asociado a la aplicación concreta. Este atributo se usará para crear una subcarpeta dentro de la carpeta  `CACHE_DIR`. De esta forma, cada aplicación tendrá una carpeta dentro de `CACHE_DIR` para almacenar sus datos.
* `cache_dir`, un string con la ruta completa de la subcarpeta que usará la aplicación para almacenar sus datos.
* `obsolescence`, un `int` que indica el número de días que consideraremos que los datos de la caché siguen siendo válidos.
  
Estos atributos serán internos (privados) y solo podrán consultarse a través de propiedades (@property), nunca modificarse desde fuera.

Deberá tener (al menos) los siguientes métodos:

* `set(self, name:str, data:str) -> None` : método que almacena en `cache_dir` los datos en `data` con el nombre `name`.
* `exists(self, name: str) -> bool` : Comprueba si ya existe un fichero con el nombre name en la caché `cache_dir`.
* `load(self, name: str) -> str`: método que recupera los datos almacenados en la caché, con el nombre `name`. Si no hay ningún dato asociado deberá lanzar la excepción `CacheError`.
* `how_old(self, name: str) -> float`: método que nos dirá la edad en milisegundos del fichero almacenado en la caché con el nombre indicado. Esto sirve para comprobar si el fichero está caducado según el valor de `obsolescence`.
* `delete(self, name: str) -> None`: método que borra de la caché el fichero  con nombre `nane`. Si  no existe, no se lanza ninguna excepción.
* `clear(self) -> None`: método que borra todos los elementos de la caché.

Se pueden definir todos los métodos auxiliares que se deseen. Si se quiere indicar algún error no indicado, se debe lanzar la excepción `CacheError`.


### Clase `CacheUrl`

La clase `CacheURL` será una especialización de la clase `Cache` para trabajar específicamente con datos descargados de internet (URLs). 

Las URLs suelen ser muy largas y contener caracteres especiales (?, =, &, / …) que pueden dar problemas al usarlas directamente como nombres de ficheros en el sistema operativo.

El objetivo principal es poder reutilizar datos de una misma URL sin necesidad de descargarla varias veces, y hacerlo de una forma segura y portable en todos los sistemas operativos.

Para evitar problemas, no vamos a guardar la URL como nombre de fichero directamente, sino que guardaremos un hash de la URL. De esta forma, se obtiene un nombre de fichero corto y seguro.

Por ejemplo:

````python
import hashlib
url = 'https://agenciatributaria.madrid.es/portal/site/contribuyente'
url_hashed = hashlib.md5(url.encode('utf-8')).hexdigest()
url_hashed
````

La clase `CacheURL` comparte el constructor con la clase `Cache` y los métodos que debe definir son:

* `get(self, url:str) -> str`
  Este método va a realizar un `get`  a la url indicada. Este método consultará si tiene ya un objeto asociado a esta petición, si no lo tiene realizará la petición correspondiente, guardará el resultado en la caché. Devolverá el objeto correspondiente.


Como el resultado depende de los argumentos dados, el nombre asociado al objeto que se guarda en la caché será un hash de la url junto con los parámetros opcionales a la petición.

Se deben redefinir los métodos de la clase de la forma adecuada:
* `exists(self, url: str, **kwargs) -> bool` 
* `load(self, url: str, **kwargs) -> bool` 
* `how_old(self, name: str, **kwargs) -> float`
* `delete(self, name: str, **kwargs) -> None`
